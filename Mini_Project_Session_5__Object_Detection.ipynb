2. Counts how many objects pass through a defined line or region

2. Imports :

```python
import os
import gdown
import cv2
import torch
from ultralytics import YOLO, solutions
from IPython.display import Video, display
```

3. Download Video :

```python
url = "https://drive.google.com/file/d/1X8oWXGBi6u_poGUvlAOD5Dj7qWs2heYL/view?usp=sharing"
video_path = "/content/bottle.mp4"

gdown.download(url, video_path, quiet=False, fuzzy=True)
assert os.path.exists(video_path)

cap = cv2.VideoCapture(video_path)
ok, _ = cap.read()
cap.release()

if not ok:
    fixed = "/content/bottle_fixed.mp4"
    !ffmpeg -y -i "{video_path}" -an -vcodec libx264 -pix_fmt yuv420p -movflags +faststart "{fixed}" -loglevel error
    video_path = fixed

print(video_path, os.path.exists(video_path))
```

4. Video Info :

```python
cap = cv2.VideoCapture(video_path)
assert cap.isOpened()

w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = float(cap.get(cv2.CAP_PROP_FPS))
if fps <= 1e-3:
    fps = 30.0

cap.release()
print(w, h, fps)
```

5. Bottle Class Id :

```python
m = YOLO("yolo11n.pt")
names = m.names
bottle_id = [k for k, v in names.items() if v == "bottle"][0]
print("BOTTLE_ID", bottle_id)
```

6. Define Region :

```python
x_line = int(w * 0.7)
region = [(x_line, 0), (x_line, h)]
print(region)
```

7. Run ObjectCounter :

```python
device = 0 if torch.cuda.is_available() else "cpu"

out_avi = "/content/object_counting_output.avi"
fourcc = cv2.VideoWriter_fourcc(*"XVID")
writer = cv2.VideoWriter(out_avi, fourcc, fps, (w, h))
assert writer.isOpened()

cap = cv2.VideoCapture(video_path)
assert cap.isOpened()

counter = solutions.ObjectCounter(
    model="yolo11n.pt",
    region=region,
    classes=[bottle_id],
    tracker="botsort.yaml",
    conf=0.10,
    iou=0.5,
    max_det=300,
    device=device,
    show=False,
    verbose=False,
)

while True:
    ok, im0 = cap.read()
    if not ok:
        break
    res = counter(im0)
    writer.write(res.plot_im)

cap.release()
writer.release()

print("IN", counter.in_count)
print("OUT", counter.out_count)
print(out_avi)
```

8. Convert And Preview :

```python
out_mp4 = "/content/object_counting_output.mp4"
!ffmpeg -y -i "{out_avi}" -an -vcodec libx264 -pix_fmt yuv420p -movflags +faststart "{out_mp4}" -loglevel error
print(out_mp4, os.path.exists(out_mp4))
display(Video(out_mp4, embed=True))
```
